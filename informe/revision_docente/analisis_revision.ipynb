# Verificación de resultados de M5
Ejecutar desde la carpeta que contiene este cuaderno. Los datos corresponden a las políticas congeladas y al reanálisis de reposición.

```python
import csv, json, statistics
from pathlib import Path
def filas(ruta):
    with Path(ruta).open(encoding='utf-8') as archivo:
        return list(csv.DictReader(archivo))
metas = filas('evidencia/meta_por_corrida.csv')
assert len(metas) == 48
for h in ('H0_0.9','H0_0.95'):
    muestra = [r for r in metas if r['referencia']==h]
    aprobadas = [r['corrida'] for r in muestra if r['cumple']=='True']
    print(h, len(aprobadas), 'de', len(muestra), 'corridas; PPO:', [r for r in aprobadas if r.startswith('P0')])
```

## Sensibilidad al precio en validación

```python
for r in filas('evidencia/sensibilidad_validacion.csv'):
    print(r['modelo'],r['politica'],round(float(r['retorno']),2),round(100*float(r['tasa_quiebres']),2),'% de quiebres')
```

## Retorno descontado del agente tabular

```python
curva = filas('evidencia/g0_tabular.csv')
assert len(curva)==900
for semilla in ('3','4','5'):
    datos = [r for r in curva if r['semilla']==semilla]
    print('Semilla',semilla,'| episodios',len(datos),'| diferencia máxima de retorno base',max(float(r['diferencia_absoluta_archivo']) for r in datos))
```
